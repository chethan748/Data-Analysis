```python

import pandas as pd 
data=pd.read_excel("../data/data.xlsx")
df=pd.DataFrame(data)
```

```python
print(df.tail(5))
print(df["Customer ID"].isnull().sum()/len(df)*100)
df["Customer ID"]=df["Customer ID"].fillna("Unknown")
print(df["Customer ID"].isnull().sum())
mask = df["Customer ID"].isnull()
df.loc[mask,"Customer ID"] =range(1001,1001+mask.sum())

df["Description"]=df["Description"].fillna(df["StockCode"])
print(df["Description"].isnull().sum())

df.drop_duplicates(inplace=True)
print("Dupplicate record ::",df.duplicated().sum())
df["InvoiceDate"]=pd.to_datetime(df["InvoiceDate"])

mask=df["Quantity"]<0
df.loc[mask,"Quantity"]=0
print((df["Quantity"]<0).sum())
print((df["Price"]<0) | (df["Price"]==0).sum())
q1=df["Quantity"].quantile(0.25)
q3=df["Quantity"].quantile(0.75)
iqr=q3-q1
lower=q1-1.5*iqr
upper=q1+1.5*iqr
outerliers=df[(df["Quantity"]<lower) | (df["Quantity"]>upper)]
q1=df["Price"].quantile(0.25)
q3=df["Price"].quantile(0.75)
iqr=q3-q1
lower=q1+1.25*iqr
upper=q3-1.25*iqr
outliers = df[(df["UnitPrice"] < lower) | (df["UnitPrice"] > upper)]

print(outliers)
df.to_excel("../data/data.xlsx", index=False)
```
