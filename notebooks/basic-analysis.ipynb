# Basic Analysis


```python

import pandas as pd 
import numpy as np
data=pd.read_excel("../data/data.xlsx")
df=pd.DataFrame(data)
print(df.head(5))
```

*What is the total revenue generated by the company?*

```python
df["TotalPrice"]=df["Quantity"]*df["Price"]
Total_Revenue=df["TotalPrice"].sum()
print("Total Revenue of the company ::",Total_Revenue)


```

*How many unique customers purchased products?*

```python
unique_c=df["Customer ID"].nunique()
print("Number of unique Coustmer ::",unique_c)
```

*How many unique products were sold*

```python
unique_p=df["StockCode"].nunique()
print("Number of unique product sold :: ",unique_p)
```

Which country has the highest number of customers?``

```python
highest_coustmers=df.groupby("Country")["Customer ID"].nunique().idxmax()
print("Highest Coustmers ::",highest_coustmers)
```

Which country has the highest number of transactions?

```python
highest_transaction=df.groupby("Country")["Invoice"].count().idxmax()
print(highest_transaction)
```

*What is the average order value?*

```python
avy_ord=df.groupby("Invoice")["TotalPrice"].sum().mean()
print("Average order value ::",avy_ord)
```

*What is the average quantity purchased per transaction?`*

```python
aqp=df.groupby("Invoice")["Quantity"].sum().mean()
print(aqp)
```
