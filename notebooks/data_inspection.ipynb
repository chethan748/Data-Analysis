```python

import pandas as pd 
data=pd.read_excel("../data/data.xlsx")
df=pd.DataFrame(data)
```

```python
print(f" First five rows :: {df.head(5)}")
print(f" Last five rows  :: {df.tail(5)}")
print(f" Number of Rows and Columns :: ",df.shape)
print(f"Columns Names ::",df.columns)
print(df.info())
print(f"Basic Statistical Information :: {df.describe()}")
print("Unique Values ::",df.nunique())
print("Total Number of Null Values in Each Column ::",df.isnull().sum())
print("Total Number of Unique Record :: ",df.duplicated().sum())
```
