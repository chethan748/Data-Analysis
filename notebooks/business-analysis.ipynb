# Business Analysis

*Find the top 10 best-selling products by quantity.*`

```python

import pandas as pd 
data=pd.read_excel("../data/data.xlsx")
df=pd.DataFrame(data)
```

```python

print("top 10 best-selling products by quantity ::",df.sort_values(by="Quantity",ascending=False).head(10))
```

*Find the top 10 products generating the highest revenue.*

```python
df["TotalPrice"]=df["Price"]*df["Quantity"]
print("top 10 products generating the highest revenue ::",df.sort_values(by="TotalPrice",ascending=False).head(10))
```

*Find the 10 least-selling products.*

```python
print("top 10 best-selling products by quantity ::",df.sort_values(by="Quantity",ascending=False).tail(10))
```

*Identify the top 10 countries by revenue.*

```python
country_revenue=df.groupby("Country")["TotalPrice"].sum()
top_countries=country_revenue.sort_values(ascending=False)
print(top_countries)
print("top 10 countries by revenue ::",top_countries.head(10))
```

*Calculate monthly revenue.*

```python
df["InvoiceDate"]=pd.to_datetime(df["InvoiceDate"])
monthly_revenue=df.groupby(df["InvoiceDate"].dt.to_period("M")["TotalPrice"].sum())
print("Monthly revenue ::",monthly_revenue)
```

*Identify the month with the highest revenue.*``

```python
df["InvoiceDate"]=pd.to_datetime(df["InvoiceDate"])
monthly_revenue=df.groupby(df["InvoiceDate"].dt.to_period("M")["TotalPrice"].sum())
print("Monthly revenue ::",monthly_revenue.sort_values(ascending=False).head(1))
```

*average spending per customer.*

```python
average_spending=df.groupby("Customer ID")["TotalPrice"].sum().mean()
print("Average Spending of per coustmer ::",average_spending)
```

*customers who generated the highest revenue*

```python
coust_rev=df.groupby("Customer ID")["TotalPrice"].sum()
coust_hig_rev=coust_rev.sort_values(ascending=False)
print("top 5 coustmer who generated highest revenue :: ",coust_hig_rev.head(5))
```
