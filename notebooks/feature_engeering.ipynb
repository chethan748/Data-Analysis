```python

import pandas as pd 
data=pd.read_excel("../data/data.xlsx")
df=pd.DataFrame(data)
```

*Create a TotalPrice column.*

```python
print(df.head(5))
df["TotalPrice"]=df["Quantity"]*df["Price"]
```

*Extract the year from InvoiceDate*

```python
year=df["InvoiceDate"].dt.year
print(year)
```

*Extract the month from InvoiceDate.*

```python
month=df["InvoiceDate"].dt.month
month
```

*Extract the day of the week.*

```python
Days_of_Week=df["InvoiceDate"].dt.dayofweek
print(Days_of_Week)
```

*Extract the hour of purchase.*

```python
hours_purchase=df["InvoiceDate"].dt.hour
```

*Create a MonthYear column (e.g. Dec-2009).*

```python
df["MonthYear"]=df["InvoiceDate"].dt.strftime('%B,%Y')
print(df["MonthYear"])
```
